Notebook ini berisi 15 latihan. Setiap soal dikerjakan dalam satu cell kode, sesuai instruksi.

```python
# Soal 1 — Buat list berisi 5 nama produk warung, lalu tampilkan jumlah datanya.
produk = ["Beras", "Telur", "Minyak", "Gula", "Mi instan"]
print("Daftar produk:", produk)
print("Jumlah produk:", len(produk))
```

```python
# Soal 2 — Tampilkan produk pertama, produk terakhir, dan produk ke-3.
print("Produk pertama:", produk[0])
print("Produk terakhir:", produk[4])
print("Produk ke-3:", produk[2])
```

```python
# Soal 3 — Tambahkan 2 produk baru dengan append(), lalu hapus 1 produk dengan remove().
produk.append("Kopi")
produk.append("Sabun")
produk.remove("Gula")
print("List setelah diperbarui:", produk)
```

```python
# Soal 4 — Buat list harga berisi 5 angka; tampilkan total, termahal, termurah, dan rata-rata.
harga = [12000, 25000, 8000, 15000, 20000]
print("Daftar harga:", harga)
print("Total harga:", sum(harga))
print("Harga termahal:", max(harga))
print("Harga termurah:", min(harga))
print("Rata-rata harga:", sum(harga) / len(harga))
```

```python
# Soal 5 — Urutkan list harga dari kecil ke besar menggunakan sort().
harga.sort()
print("Harga dari kecil ke besar:", harga)
```

```python
# Soal 6 — Ambil 3 produk pertama menggunakan potongan list (produk[0:3]).
tiga_produk_pertama = produk[0:3]
print("Tiga produk pertama:", tiga_produk_pertama)
```

```python
# Soal 7 — Gabungkan list produk menjadi satu kalimat dengan ' - '.join(produk).
kalimat_produk = " - ".join(produk)
print(kalimat_produk)
```

```python
# Soal 8 — Buat dictionary pelanggan berisi nama, kota, no_hp, dan total_belanja.
pelanggan = {
    "nama": "Aisyah Putri",
    "kota": "Mataram",
    "no_hp": "081234567890",
    "total_belanja": 25000
}
print(pelanggan)
```

```python
# Soal 9 — Ubah total_belanja menjadi dua kali lipat, lalu tampilkan hasilnya.
pelanggan["total_belanja"] *= 2
print("Total belanja setelah dikali dua:", pelanggan["total_belanja"])
```

```python
# Soal 10 — Tambahkan kunci member bernilai True, lalu hapus kunci no_hp.
pelanggan["member"] = True
del pelanggan["no_hp"]
print(pelanggan)
```

```python
# Soal 11 — Tampilkan semua kunci dan semua nilai dictionary tersebut.
print("Semua kunci:", list(pelanggan.keys()))
print("Semua nilai:", list(pelanggan.values()))
```

```python
# Soal 12 — Coba ambil kunci 'diskon' dengan get() agar tidak error.
diskon = pelanggan.get("diskon", "Kunci diskon belum tersedia")
print("Diskon:", diskon)
```

```python
# Soal 13 — Buat list berisi 3 dictionary produk (nama, harga, stok), lalu tampilkan nama produk kedua.
daftar_produk = [
    {"nama": "Beras", "harga": 14000, "stok": 20},
    {"nama": "Telur", "harga": 28000, "stok": 15},
    {"nama": "Minyak", "harga": 18000, "stok": 10}
]
print("Nama produk kedua:", daftar_produk[1]["nama"])
```

```python
# Soal 14 — Tanyakan nama, kota, dan 3 produk favorit lewat input(), lalu tampilkan ringkasan berbentuk kartu.
nama = input("Masukkan nama: ").strip().title()
kota = input("Masukkan kota: ").strip().title()
produk_favorit = []
for i in range(3):
    item = input(f"Masukkan produk favorit ke-{i+1}: ").strip().title()
    produk_favorit.append(item)

print("\n" + "=" * 32)
print("       KARTU PROFIL PELANGGAN")
print("=" * 32)
print(f"Nama            : {nama}")
print(f"Kota            : {kota}")
print(f"Produk favorit  : {', '.join(produk_favorit)}")
print("=" * 32)
```

```python
# Soal 15 — Rapikan data mentah menjadi dictionary bersih.
# Data mentah: nama dan kota dirapikan, harga diubah menjadi angka.
data_mentah = "  aisyah PUTRI , mataram , 25000  "
bagian = [item.strip() for item in data_mentah.split(",")]
nama_bersih = bagian[0].title()
kota_bersih = bagian[1].title()
harga_bersih = int(bagian[2])
data_bersih = {
    "nama": nama_bersih,
    "kota": kota_bersih,
    "harga": harga_bersih
}
print("Dictionary bersih:", data_bersih)
print("Nama:", data_bersih["nama"])
print("Kota:", data_bersih["kota"])
print("Harga:", data_bersih["harga"], "(tipe:", type(data_bersih["harga"]).__name__ + ")")
```
